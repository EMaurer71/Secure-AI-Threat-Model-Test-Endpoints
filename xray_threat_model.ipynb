# Threat Modeling — Medical Diagnostic AI (Chest X‑Ray Analysis)

System: AI system that analyzes chest X‑rays to detect pneumonia, tuberculosis, and other respiratory conditions.

Methodology: STRIDE-based threat modeling for AI-enabled medical diagnostics.

## 1. System Overview

### 1.1 System Configuration
**Medical Diagnostic AI (X-Ray Analysis)**

AI system that analyzes chest X‑rays to detect pneumonia, tuberculosis, and other respiratory conditions.

### 1.2 Critical Assets
- Patient health data
- Diagnostic model
- Medical images
- Doctor recommendations

### 1.3 Attack Surface
- API endpoints
- Training data pipeline
- Model inference engine
- Output interface

## 2. Data Flow & Trust Boundaries

### 2.1 High-Level Data Flow
1. Clinician workstation captures or uploads chest X‑ray.
2. Image upload API receives the image and metadata.
3. Preprocessing pipeline normalizes, validates, and transforms the image.
4. Diagnostic model performs inference.
5. Results returned to clinician UI.
6. Reports stored in EHR.

### 2.2 Trust Boundaries
- Boundary A: External network → Hospital network
- Boundary B: Hospital network → AI inference service
- Boundary C: AI service → EHR systems
- Boundary D: Training data sources → Model training environment

## 3. STRIDE Summary

STRIDE categories:
- **Spoofing**
- **Tampering**
- **Repudiation**
- **Information Disclosure**
- **Denial of Service**
- **Elevation of Privilege**

## 4. STRIDE Analysis by Component

### 4.1 API Endpoints
**Spoofing:** Attacker spoofs clinician identity.

**Tampering:** X‑ray images or metadata modified.

**Repudiation:** Clinician denies submitting an X‑ray.

**Information Disclosure:** API leaks patient data.

**Denial of Service:** Flooding inference API.

**Elevation of Privilege:** Misconfig allows admin access.

### 4.2 Training Data Pipeline
**Spoofing:** Attacker impersonates trusted data source.

**Tampering:** Labels or images altered (data poisoning).

**Repudiation:** Data provider denies dataset.

**Information Disclosure:** Training data leaks PHI.

**Denial of Service:** ETL or training jobs disrupted.

**Elevation of Privilege:** Attacker gains training environment access.

### 4.3 Model Inference Engine
**Spoofing:** Fake inference service returns bogus results.

**Tampering:** Model weights modified.

**Repudiation:** No traceability to model version.

**Information Disclosure:** Logs leak PHI.

**Denial of Service:** Heavy inputs overload engine.

**Elevation of Privilege:** Runtime exploit → host access.

### 4.4 Output Interface
**Spoofing:** Phishing UI mimics dashboard.

**Tampering:** Reports altered before EHR.

**Repudiation:** Clinician denies viewing diagnosis.

**Information Disclosure:** UI exposes excessive data.

**Denial of Service:** Dashboard unavailable.

**Elevation of Privilege:** UI vulnerability escalates privileges.

## 6. Example Input Validation Code
Basic structural validation for X-ray images.

````python
import numpy as np

def validate_xray(image: np.ndarray):
    if not isinstance(image, np.ndarray):
        raise ValueError("Input must be a NumPy array.")
    if image.ndim != 2:
        raise ValueError("X-ray must be a 2D grayscale image.")
    if image.shape != (128, 128):
        raise ValueError("Expected 128x128 image.")
    if image.min() < 0 or image.max() > 255:
        raise ValueError("Pixel values out of range.")
    return True

validate_xray(np.zeros((128,128)))
````
